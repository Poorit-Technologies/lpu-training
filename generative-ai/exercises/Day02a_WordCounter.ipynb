**When it misbehaves:**

| What you see | What it means |
|---|---|
| `NameError: name 'gr' is not defined` | you skipped the setup cells — run the notebook from the top |
| The app renders but the button does nothing | your function doesn't `return` anything |
| Nothing renders below the cell | you forgot `.launch()` |
| `TypeError` about arguments | the number of `inputs` must match the number of arguments your function takes |
