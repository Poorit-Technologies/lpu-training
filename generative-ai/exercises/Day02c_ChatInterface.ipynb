**When it misbehaves:**

| What you see | What it means |
|---|---|
| `AuthenticationError` | the environment variable never got set — re-run your Step 1 cell |
| The bot forgets what you just said | you never passed `history` into the message list — Gradio hands it to you, you have to use it |
| `history` arrives as lists, not dicts | you left out `type="messages"` |
| Nothing renders below the cell | you forgot `.launch()` |
| `BadRequestError: LLM Provider NOT provided` | the model string needs its prefix: `openai/gpt-4o-mini`, not `gpt-4o-mini` |

### ✅ What you practised

| Idea | The one-liner |
|---|---|
| **`gr.ChatInterface(fn, type="messages")`** | a whole chat UI from one function |
| **`fn(message, history)`** | Gradio stores the conversation and hands it back every turn |
| **Memory is just messages** | the model is stateless — the history you resend *is* the memory |
| **LiteLLM `completion()`** | swap the model string to change provider, nothing else moves |
| **`yield` instead of `return`** | the reply types out live instead of landing all at once |
