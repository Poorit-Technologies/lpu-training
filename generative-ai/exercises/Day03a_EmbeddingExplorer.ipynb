---

**When it misbehaves:**

| What you see | What it means |
|---|---|
| `NameError: name 'model' is not defined` | you skipped the setup cells — run from the top |
| `ValueError: Expected 2D array, got 1D` | `encode("one string")` gives a 1-D vector; pass a **list**, even for one item |
| The plot is a single blob | your sentences are all about the same thing — make the topics further apart |
| `c` argument error in `scatter` | your colour list is a different length from your sentence list |
| Every similarity is ~0.99 | you probably embedded the same text repeatedly — check your list |

### ✅ What you practised

| Idea | The one-liner |
|---|---|
| **`model.encode(texts)`** | a list of texts → an array of `(n, 384)` |
| **`cosine_similarity`** | compares every vector against every other in one call |
| **PCA** | squashes 384 dimensions to 2 so a human can look at it — a shadow, not the truth |
| **Nearest neighbours** | "search" is just: embed the query, rank by similarity, take the top k |
